## 1. My lane as an ML task (type)
**Selected Lane:** Refresh / Content Opportunity Scoring

**Machine Learning Task Type:** Scoring

The goal of this project is to assign a refresh priority score to each webpage based on its content quality and performance metrics. Pages with higher scores should be reviewed first because they are more likely to benefit from a content update. A scoring model is appropriate because it produces a numerical value that can be used to prioritize pages instead of making only a yes/no decision.


## 2. Target or Proxy

The target is a **Content Refresh Priority Score**.

Since the dataset is anonymized and may not contain a direct label showing whether a page needed updating, I will use a proxy based on the available performance features. Pages with lower performance or lower content quality indicators can be assigned a higher refresh priority. This is a decision-support score rather than a guaranteed prediction.


```python
import pandas as pd

data = pd.read_csv("/content/content_refresh_anonymized.csv")

data.shape[0] , data.shape[1] #Rows & Columns
```

## 3. Success Metric

The primary success metric is **Precision@50**.

This measures how many of the top 50 pages recommended by the model are actually good candidates for content refresh. A higher Precision@50 means the model helps prioritize the most valuable pages for review, making the recommendations more useful for decision-making.

## 4. The unit of analysis, as a real dataframe

The unit of analysis is **one webpage**.

Each row in the dataset represents a single webpage, and each column contains measured characteristics such as performance, freshness, or quality-related features. The model will use these features to estimate a refresh priority score for each page.


```python
data.head()

print(data.columns.tolist())
```

## 5. Why ML beats a fixed rule here

A fixed rule, such as "refresh every page older than one year," ignores many other important factors. Different pages may have different traffic patterns, content quality, and engagement levels.

Machine Learning can evaluate multiple features together and learn relationships between them. This allows it to produce more accurate and flexible recommendations than a single manually written rule. The model provides decision-support rather than replacing human judgment.
